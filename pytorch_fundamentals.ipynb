# reshaping,view, stacking (vstack,hstack),permute, squeezing and unsqueezin tensors



```python
# create a tensor

import torch
x=torch.arange(1.,10.)
x,x.shape
```

```python
# add an extra dimension
x_reshaped=x.reshape(1,9)
x_reshaped,x_reshaped.ndim,x_reshaped.shape
```

```python
# add an extra dimension
x_reshaped1=x.reshape(9,1)
x_reshaped1,x_reshaped1.ndim,x_reshaped1.shape
```

```python
# add an extra dimension
x_reshaped=x.reshape(5,2)
x_reshaped,x_reshaped.ndim,x_reshaped.shape
```

```python
# change the view
z=x.view(1,9)
z,z.shape

# changes z changes x (becouse view of a tensor shares the same mamory as the original input)
```

```python
z[:,0]=5
z,x
```

```python
# stack tensors on top of each other
x_stacked=torch.stack([x,x,x,x],dim=0)
x_stacked
```

```python
x_stacked=torch.stack([x,x,x,x],dim=1)
x_stacked
```

# Reproducbility (trying to take random out of random)

```python
tensor_a=torch.rand(3,4)
tensor_b=torch.rand(3,4)

print(tensor_a)
print(tensor_b)
print(tensor_a==tensor_b)
```

```python
# let's make some random but reproducible tensors
import torch

# set the rendom seed

RANDOM_SEED=42
torch.manual_seed(RANDOM_SEED)
random_tensor_c=torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
random_tensor_d=torch.rand(3,4)
# torch.manual_seed(RANDOM_SEED)

print(random_tensor_c)
print(random_tensor_d)
print(random_tensor_c==random_tensor_d)


```
